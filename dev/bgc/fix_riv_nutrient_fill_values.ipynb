```python
import xarray as xr
from netCDF4 import default_fillvals
```

```python
ds = xr.open_dataset("/glade/campaign/cesm/cesmdata/inputdata/ocn/mom/croc/testing_data/panama-bgc/river_nutrients_panama1_5490e0_nnsm.nc")
```

```python
ds["nx"] = ds.nx
ds["nx"].attrs["cartesian_axis"] = "X"
ds["ny"] = ds.ny
ds["ny"].attrs["cartesian_axis"] = "Y"
```

```python
fv = default_fillvals["f8"]
ds.to_netcdf(
    "river_nutrients_panama1_5490e0_nnsm.nc",
    encoding={var: {"_FillValue": fv} for var in ds.data_vars}
)
```
