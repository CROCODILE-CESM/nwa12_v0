```python
import xarray as xr 
import numpy as np 
import re
import os
```

```python
datasets = []
datasets.append(xr.open_mfdataset("/glade/u/home/manishrv/documents/croc/dev/bgc/old_format_files/forcing_obc_segment_001.nc"))
datasets.append(xr.open_mfdataset("/glade/u/home/manishrv/documents/croc/dev/bgc/old_format_files/forcing_obc_segment_002.nc"))
datasets.append(xr.open_mfdataset("/glade/u/home/manishrv/documents/croc/dev/bgc/old_format_files/forcing_obc_segment_003.nc"))
datasets.append(xr.open_mfdataset("/glade/u/home/manishrv/documents/croc/dev/bgc/old_format_files/forcing_obc_segment_004.nc"))
datasets[0].temp_segment_001.dims
```

```python
# Regex to capture base variable name: "temp" from "temp_segment_001"
pattern = re.compile(r"(.+)_segment_\d+")

# Get base variable names from first dataset
base_vars = []

for var in datasets[0].data_vars:
    m = pattern.match(var)
    if m and "dz" not in m.group(1):
        base_vars.append(m.group(1))  # "temp"
```

```python
bgc_vars =[]
for var in base_vars:
    if var not in ["temp", "salt","u","v", "eta"]:
        bgc_vars.append(var)

bgc_vars
```

```python
for var in base_vars:
    ds_complete = xr.Dataset()
    for i, ds in enumerate(datasets, start=1):
        seg_suffix = f"_segment_{i:03d}"  # 001, 002, 003, 004
        ds_complete[var+seg_suffix] = ds[var+seg_suffix]
        if "eta" not in var:
            ds_complete["dz_"+var+seg_suffix] = ds["dz_"+var+seg_suffix]
    ds_complete.to_netcdf("reformatted_files/"+var+"_obc_segment.nc",unlimited_dims="time")
    print("Finished", var+"!")
```

```python
for i, ds in enumerate(datasets, start=1):
    for var in bgc_vars:
            vars = [f"{var}_segment_00{i}",f"dz_{var}_segment_00{i}",f"nz_segment_00{i}_{var}"]
            ds = ds.drop_vars(vars)
            print("Dropped", vars)
    ds.to_netcdf(f"reformatted_files/forcing_obc_segment_00{i}.nc",unlimited_dims="time")
```
