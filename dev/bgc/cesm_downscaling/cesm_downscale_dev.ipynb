```python
# Find all files that have the tracer
import re
date_format = "%Y%m"
files = {}
for tracer in tracers:
    files[tracer] = {"file_paths":list(Path(cesm_output_dir).rglob(f"*.h.{tracer}.*.nc"))}

for tracer in tracers:
    earliest_date = None
    latest_date = None
    for file in files[tracer]["file_paths"]:
        match = re.search(r"\.(\d{6})-(\d{6})\.nc$", str(file))
        if match:
            start_date = datetime.strptime(match.group(1), date_format)
            end_date = datetime.strptime(match.group(2), date_format)

            # Update earliest and latest dates
            if earliest_date is None or start_date < earliest_date:
                earliest_date = start_date
            if latest_date is None or end_date > latest_date:
                latest_date = end_date
    files[tracer]["earliest_date"] = earliest_date
    files[tracer]["latest_date"] = latest_date
```

```python
# Take a given date range, and build a list of files to grab from the dict, but first check the range is in the earliest date and latest date
start_date = datetime.strptime("20000101", date_format+"%d")
end_date = datetime.strptime("20001201", date_format+"%d")
filtered_files = {}
for tracer in tracers:
    if start_date < files[tracer]["earliest_date"] or end_date > files[tracer]["latest_date"]:
        raise ValueError(f"Invalid dates for tracer: {tracer}")
    # Filter files that overlap with the given date range
    filtered_files[tracer] = []
    for file in files[tracer]["file_paths"]:
        match = re.search(r"\.(\d{6})-(\d{6})\.nc$", str(file))
        if match:
            file_start_date = datetime.strptime(match.group(1), date_format)
            file_end_date = datetime.strptime(match.group(2), date_format)

            # Check if the file overlaps with the given date range
            if file_start_date <= end_date and file_end_date >= start_date:
                filtered_files[tracer].append(str(file))
```

```python
from regional_mom6 import regridding as rgd

segment="north"
tracer = "PO4"
boundary_number_conversion = {
    "south": 1,
    "north": 2,
    "east": 3,
    "west": 4
}
hgrid = xr.open_dataset("/glade/derecho/scratch/manishrv/croc_input/nwa_in_crocesm/ocnice/ocean_hgrid.nc")
z_dim = "z_t"
coords = rgd.coords(hgrid, segment, "segment_{:03d}".format(boundary_number_conversion[segment]))
regridded_folder = Path("/glade/u/home/manishrv/documents/croc/dev/bgc/cesm_downscaling/regridded_data")
ds = xr.open_dataset(regridded_folder/f"{tracer}_{segment}_regridded.nc")
segment_name = "segment_{:03d}".format(boundary_number_conversion[segment])
dim_name = ds.variables["__xarray_dataarray_variable__"].dims[-1]
# First reshape err in regrid
if segment in ["east","west"]:
    ds.variables["__xarray_dataarray_variable__"].dims = ('time', z_dim, dim_name, "dummy")
    ds[tracer] = ds.__xarray_dataarray_variable__[:,:,:,0]
else:
    ds.variables["__xarray_dataarray_variable__"].dims = ('time', z_dim, 'dummy', dim_name)
    ds[tracer] = ds.__xarray_dataarray_variable__[:,:,0,:]
ds = ds.drop_vars(["__xarray_dataarray_variable__"])
ds
```

```python
ds = ds.rename(
            {"lon": f"lon_{segment_name}", "lat": f"lat_{segment_name}"}
        )
ds[tracer] = rgd.fill_missing_data(
            ds[tracer],
            xdim=dim_name,
            zdim=z_dim,
        )
```

```python
# Since CESM , don't need to modify time dim
ds.time
```

```python
v = f"{tracer}_{segment_name}"
            ## Rename each variable in dataset
ds = ds.rename({tracer: v})

ds = rgd.vertical_coordinate_encoding(
    ds, v, segment_name, z_dim
)

ds = rgd.add_secondary_dimension(
    ds, v, coords, segment_name
)

ds = rgd.generate_layer_thickness(
    ds, v, segment_name, z_dim
)
ds
```

```python
# Overwrite the actual lat/lon values in the dimensions, replace with incrementing integers
import numpy as np
ds[f"{coords.attrs['parallel']}_{segment_name}"] = np.arange(
    ds[f"{coords.attrs['parallel']}_{segment_name}"].size
)
ds[f"{coords.attrs['parallel']}_{segment_name}"] = np.arange(
    ds[f"{coords.attrs['parallel']}_{segment_name}"].size
)
ds[f"{coords.attrs['perpendicular']}_{segment_name}"] = [0]
encoding_dict = {
    "time": {"dtype": "double"},
    f"nx_{segment_name}": {
        "dtype": "int32",
    },
    f"ny_{segment_name}": {
        "dtype": "int32",
    },
}
ds = rgd.mask_dataset(
    ds,
    hgrid,
    None,
    segment,
    segment_name,
)
encoding_dict = rgd.generate_encoding(
    ds,
    encoding_dict,
    default_fill_value=1.0e20,
)
ds
```
