```python
import xarray as xr 
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
old_dir = Path("/glade/u/home/manishrv/scratch/croc_input/sauron.cesm.4.old/ocnice")
new_dir = Path("/glade/u/home/manishrv/scratch/croc_input/sauron.cesm.3/ocnice")
```

```python

def analyze_array(old,new):
    print(old.shape,new.shape)
    # Mask NaNs
    mask_ignore =  np.isclose(old, -1e20, rtol=1e-6, atol=0) | np.isclose(new, -1e20, rtol=1e-6, atol=0)

    # Compare the rest
    diff_mask = ~np.isclose(old, new, rtol=1e-6, atol=0)
    # Ignore NaNs
    diff_mask[mask_ignore] = False
    return np.where(diff_mask)
```

```python

boundaries_old = {}
boundaries_new = {}
for s in ["001","002","003","004"]:
    boundaries_old[s] = xr.open_mfdataset(str(old_dir/f"*_obc_segment_{s}.nc"))
    boundaries_new[s] = xr.open_dataset(new_dir/f"forcing_obc_segment_{s}.nc")
```

```python
exclude = ["SSH", "UVEL", "VVEL", "TEMP", "SALT", "dz"]

for s in ["001","002","003","004"]:
    for var in boundaries_old[s].data_vars:
        
        if not any(exc in var for exc in exclude):
            print(var)
            print(analyze_array(boundaries_old[s][var],boundaries_new[s][var]))


    print(analyze_array(boundaries_old[s][f"SALT_segment_{s}"],boundaries_new[s][f"salt_segment_{s}"]))
    print(analyze_array(boundaries_old[s][f"TEMP_segment_{s}"],boundaries_new[s][f"temp_segment_{s}"]))
    print(analyze_array(boundaries_old[s][f"SSH_segment_{s}"],boundaries_new[s][f"eta_segment_{s}"]/100))
    print(analyze_array(boundaries_old[s][f"UVEL_segment_{s}"],boundaries_new[s][f"u_segment_{s}"]/100))
    print(analyze_array(boundaries_old[s][f"VVEL_segment_{s}"],boundaries_new[s][f"v_segment_{s}"]/100))

```
