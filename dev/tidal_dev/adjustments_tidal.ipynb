```python
directory = "/glade/u/home/manishrv/documents/nwa12_0.1/regional_mom_workflows/crr/tides_dev/outputs/forcing"
output_directory = "/glade/u/home/manishrv/documents/nwa12_0.1/regional_mom_workflows/crr/tides_dev/outputs/nonasforcing"

# Iterate through the files in the directory
for filename in os.listdir(directory):
    if filename.startswith("tu") or filename.startswith("tz"):
        filepath = os.path.join(directory, filename)
        output_filepath = os.path.join(output_directory, filename)

        print(filepath)
        # Open the NetCDF file
        ds = xr.open_dataset(filepath)
        
        # Replace NaNs with zeros in all data variables
        ds = ds.fillna(0)

        ds.to_netcdf(output_filepath,
        unlimited_dims="time")
```
