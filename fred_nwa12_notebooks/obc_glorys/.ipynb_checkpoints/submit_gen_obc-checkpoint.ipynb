# create and submit a pbs job to create open boundary conditions:


```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from mom6_carib_routines import *
from datetime import datetime
from pathlib import Path

%matplotlib inline
```

```python
def make_PBS_batch(year,noformat_out,format_out,script_out,proj_code='P93300612'):
  """
  Create and submit a PBS batch script to run gen_obc_glorys2mom6.py
  """
  job_name = "{}_obc".format(year)
  print('Writing scripts/gen_{}_obc.sh...'.format(year))
  fname = '{}gen_{}_obc.sh'.format(script_out, year)
  f = open(fname, 'w')
  f.write("#!/bin/bash \n")
  f.write("#PBS -N {} \n".format(job_name))
  f.write("#PBS -A {} \n".format(proj_code))
  f.write("#PBS -l select=1:ncpus=1:mem=100GB \n")
  f.write("#PBS -l walltime=12:00:00 \n")
  f.write("#PBS -q casper \n")
  f.write("#PBS -o {} \n".format(script_out))
  f.write("#PBS -j oe \n")
  f.write("    \n")
  f.write("source /glade/u/home/gseijo/miniconda3/etc/profile.d/conda.sh      \n")
  f.write("conda activate /glade/u/home/gseijo/miniconda3/envs/gridTools  \n")
  f.write("    \n")
  f.write("python /glade/work/gseijo/python_codes/gen_glorys_obc_v2.py {} {} {}".format(year,noformat_out, format_out))
  f.close()

  os.system('qsub {}'.format(fname))
    
  return
```

```python
years = np.arange(2005,2022)
script_out = '/glade/scratch/gseijo/glorys012/obc/scripts/'
make_dir(script_out)
for year in years:
    noformat_out = '/glade/scratch/gseijo/glorys012/obc/'+str(year)+'/'
    format_out   = '/glade/work/gseijo/Carib12-test/inputs/obc/'+str(year)+'/'
    make_PBS_batch(year,noformat_out,format_out,script_out, proj_code = 'P93300612')
```
