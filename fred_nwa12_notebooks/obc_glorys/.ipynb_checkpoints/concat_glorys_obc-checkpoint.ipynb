```python
import xarray as xr
import sys
sys.path.append("/glade/work/gseijo/python_codes/") 
from mom6_carib_routines import *


#parent input dir:
gen_dir_in = '/glade/work/gseijo/Carib12-test/inputs/obc/'
#output dir:
dir_out    = '/glade/work/gseijo/Carib12-test/inputs/obc/2000_2020/'
#make sure output dir exists:
make_dir(dir_out)
#year to concat:
years = np.arange(1999, 2022)
#segment names:
bnds  = ['001','002']
#variables to concat:
data  = ['salt','temp','u','v','zeta']

for j in range(1):#len(bnds)):
    for k in range(1):#len(data)):
        for i in range(len(years)):
                print('opening: '+ gen_dir_in+str(years[i])+'/combined_'+str(data[k])+'_'+str(bnds[j])+'_'+str(years[i])+'.nc')
                    
                if i==0:
                        vars()[data[k]+'_'+bnds[j]] = xr.open_dataset(gen_dir_in+str(years[i])+'/combined_'+str(data[k])+'_'+str(bnds[j])+'_'+str(years[i])+'.nc',decode_times=False)
                elif i!=0:
                        print('concatenating: '+ gen_dir_in + str(years[i])+'/combined_'+str(data[k])+'_'+str(bnds[j])+'_'+str(years[i])+'.nc')
                        bfr = xr.open_dataset(gen_dir_in+str(years[i])+'/combined_'+str(data[k])+'_'+str(bnds[j])+'_'+str(years[i])+'.nc',decode_times=False)
                        
                        if years[i] == 2021:                           
                            bfr2 = xr.Dataset(coords={'time': bfr.time.values + 24})
                            bfr = bfr.assign_coords(time=bfr2['time'])
                            
                        vars()[data[k]+'_'+bnds[j]] = xr.concat([vars()[data[k]+'_'+bnds[j]],bfr],dim='time')
                                                                       
                if i==len(years)-1:
                    #make sure atributes are saved correctly:
                    for v in vars()[data[k]+'_'+bnds[j]]:
                           vars()[data[k]+'_'+bnds[j]][v].encoding['_FillValue']=1.e20 
                    vars()[data[k]+'_'+bnds[j]].time.attrs['calendar']='gregorian'
                    vars()[data[k]+'_'+bnds[j]].time.attrs['units']='hours since 1950-01-01'
                    vars()[data[k]+'_'+bnds[j]].to_netcdf(dir_out+str(data[k])+'_'+bnds[j]+'.nc',unlimited_dims='time')
                    print('saved: ' + dir_out+str(data[k])+'_'+str(bnds[j])+'.nc')                                      
```
